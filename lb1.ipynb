```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt
```

```python
def adjusted_contrast(image, alpha, beta):
    adjusted_image = cv2.convertScaleAbs(image, alpha = alpha, beta = beta)
    return adjusted_image
```

```python
def equalize_histogram(image):
    equalize_image = cv2.equalizeHist(image)
    return equalize_image
```

```python
image_input = cv2.imread('./images/tiger.jpg', cv2.IMREAD_GRAYSCALE)
alpha = 1.5
beta = 0
```

```python
contrast_adjusted_image = adjusted_contrast(image_input, alpha, beta)
equalize_image = equalize_histogram(image_input)
```

```python
cv2.imshow('Original image', image_input)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imshow('Contrast adjusted image', contrast_adjusted_image)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imshow('equalized image', equalize_image)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imwrite('contrast_adjusted_op.jpg', contrast_adjusted_image)
cv2.imwrite('equalized_op.jpg', equalize_image)
```

```python
plt.figure(figsize = (12, 6))
plt.subplot(2, 2, 1)
plt.hist(image_input.ravel(), 256, [0, 256])
plt.title('original image histogram')
plt.subplot(2, 2, 2)
plt.hist(contrast_adjusted_image.ravel(), 256, [0, 256])
plt.title('contrast adjusted image histogram')
plt.subplot(2, 2, 3)
plt.hist(equalize_image.ravel(), 256, [0, 256])
plt.title('equalized image histogram')
plt.show()
```
