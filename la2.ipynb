# resize crop and rotate

```python
import cv2
import numpy as np
```

```python

image_path = './images/tiger.jpg'
image = cv2.imread(image_path)
```

```python

resized_image = cv2.resize(image, (800, 500))
h, w, _ = resized_image.shape

start_x = (w - 200) // 2
start_y = (h - 200) // 2
```

```python

cropped_image = resized_image[start_y:start_y+200, start_x:start_x+200]

(h, w) = cropped_image.shape[:2]
center = (w//2, h//2)
```

```python

rotation_matrix = cv2.getRotationMatrix2D(center, 90, 1.0)
rotated_image = cv2.warpAffine(cropped_image, rotation_matrix, (w, h))
```

```python

cv2.imshow('Original image', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Resized image', resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Cropped image', cropped_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Rotated image', rotated_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
