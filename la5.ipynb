```python
import numpy as np
import cv2

# Create a black image
img = np.zeros((510, 510, 3), dtype='uint8')
```

```python
img = cv2.line(
    img,
    (0, 0),
    (100, 100),
    (255, 255, 255),
    4
)
```

```python
img = cv2.arrowedLine(
    img,
    (0, 50),
    (100, 150),
    (255, 255, 255),
    4
)
```

```python
img = cv2.ellipse(
    img,
    (300, 100),
    (100, 50),
    30,
    0,
    360,
    (255, 0, 0),
    -1
)
```

```python
img = cv2.circle(
    img,
    (450, 100),
    30,
    (255, 0, 0),
    4
)
```

```python
img = cv2.rectangle(
    img,
    (20, 200),
    (200, 300),
    (0, 255, 0),
    2
)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX
org = (50, 400)
fontScale = 2
color = (0, 0, 255)
thickness = 3
```

```python
img = cv2.putText(
    img,
    'OpenCV',
    org,
    font,
    fontScale,
    color,
    thickness,
    cv2.LINE_AA,
    False
)
```

```python
img = cv2.putText(
    img,
    'OpenCV',
    org,
    font,
    fontScale,
    color,
    thickness,
    cv2.LINE_AA,
    True
)
```

```python
cv2.imshow("Geometric Shapes", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
