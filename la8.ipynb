# Edge detection

```python
import cv2
image = cv2.imread('./images/tiger.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (3,3), 0)
```

```python
wide_edge = cv2.Canny(blurred, 10, 100)
mid_edge = cv2.Canny(blurred, 30, 150)
tight_edge = cv2.Canny(blurred, 240, 250)
```

```python
cv2.imshow("original image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow("wide edge image", wide_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow("mid edge image", mid_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow("tight edge image", tight_edge)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# counter detection

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (3,3), 0)
```

```python
wide_edge = cv2.Canny(blurred, 10, 100)
mid_edge = cv2.Canny(blurred, 30, 150)
tight_edge = cv2.Canny(blurred, 240, 250)
```

```python
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (3,3))
dilated = cv2.dilate(mid_edge, kernel, iterations = 1)
```

```python
contours, _ = cv2.findContours(dilated, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
image_copy = image.copy()
```

```python
cv2.drawContours(image_copy, contours, -1, (0, 255, 0), 2)
print(len(contours), "objects were found in this image")
```

```python
cv2.imshow("contours", image_copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
