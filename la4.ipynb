```python
import cv2
import numpy as np
```

```python
img1 = cv2.imread('./images/bitwise1.jpg')
img2 = cv2.imread('./images/bitwise2.jpg')
```

```python

dest_and = cv2.bitwise_and(img2, img1)
dest_or = cv2.bitwise_or(img2, img1)
dest_not1 = cv2.bitwise_not(img1)
dest_not2 = cv2.bitwise_not(img2)
dest_xor = cv2.bitwise_xor(img2, img1)
```

```python

cv2.imshow('Original image 1', img1)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Original image 2', img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Bitwise AND', dest_and)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Bitwise OR', dest_or)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Bitwise NOT 1', dest_not1)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Bitwise NOT 2', dest_not2)
cv2.waitKey(0)
cv2.destroyAllWindows()
cv2.imshow('Bitwise XOR', dest_xor)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
